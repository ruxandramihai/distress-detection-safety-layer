```python
import json
import pandas as pd
import random
import math
random.seed(42)
```

```python
def load_reddit_2023_highrisk(csv_file):
    try:
        df = pd.read_csv(csv_file)
        print(f"CSV columns in {csv_file}: {df.columns.tolist()}")
        
        unnamed_cols = [col for col in df.columns if 'Unnamed:' in col]
        if unnamed_cols:
            df = df.drop(columns=unnamed_cols)
            print(f"Dropped {len(unnamed_cols)} unnamed column(s)")
        
        required_columns = ['Title', 'Post', 'Label']
        missing_columns = [col for col in required_columns if col not in df.columns]
        if missing_columns:
            print(f"Error: Required column(s) {missing_columns} not found in {csv_file}.")
            print(f"Available columns: {df.columns.tolist()}")
            return None
        
        original_count = len(df)
        df = df[df['Label'] == 'Suicidal']
        print(f"Filtered from {original_count} to {len(df)} records with 'Suicidal' label")

        df = df.rename(columns={'Title': 'title', 'Post': 'text'})
        print("Renamed 'Title' to 'title' and 'Post' to 'text'")
        
        df = df[['title', 'text']]
        
        print(f"Successfully loaded {len(df)} 'Suicidal' records from {csv_file}")
        return df
        
    except Exception as e:
        print(f"Error loading CSV file {csv_file}: {e}")
        return None
```

```python
dataset_high_risk = load_reddit_2023_highrisk("../data/raw/Suicidal Ideation Detection reddit Dataset.csv")
```

Convert to json 

```python
def convert_to_json_with_high_risk(dataset, origin_name="suicide_detection_2023_reddit"):
    json_data = {
        "origin": origin_name,
        "data": []
    }
    
    for _, row in dataset.iterrows():
        entry = {
            "text": row["text"],
            "my_label": "high-risk"
        }
        json_data["data"].append(entry)
    
    print(f"Created JSON data with {len(json_data['data'])} entries, all labeled as 'high-risk'")
    return json_data
```

```python
json_dataset_high_risk = convert_to_json_with_high_risk(dataset_high_risk
)
```

Save json file 

```python
def save_json_to_file(json_data, output_file):
    try:
        with open(output_file, 'w', encoding='utf-8') as f:
            json.dump(json_data, f, ensure_ascii=False, indent=2)
        print(f"Successfully saved JSON data to {output_file}")
    except Exception as e:
        print(f"Error saving JSON data: {e}")
```
